## 2. NumPy Fundamentals

```python
import numpy as np

# Load and save
# arr = np.loadtxt('file.txt')
# arr = np.genfromtxt('file.csv', delimiter=',', skip_header=1)
# np.savetxt('output.csv', arr, delimiter=',')

# Create arrays
a = np.array([1, 2, 3])                 # 1D array from a list
b = np.array([[1, 2], [3, 4]])          # 2D array (matrix) from nested lists
np.zeros(3)                             # array of zeros, shape (3,)
np.ones((3, 4))                         # array of ones, shape (3, 4)
np.eye(5)                               # 5x5 identity matrix
np.full((2, 3), 8)                      # 2x3 array filled with the value 8
np.arange(0, 10, 3)                     # evenly spaced integers 0,3,6,9 (like range)
np.linspace(0, 100, 6)                  # 6 evenly spaced floats from 0 to 100 (inclusive)

# Random numbers
np.random.rand(4, 5)                    # uniform draws in [0, 1), shape (4, 5)
np.random.randint(0, 5, size=(2, 3))    # random integers in [0, 5), shape (2, 3)

# Probability distributions (modern generator API, reproducible via seed)
rng = np.random.default_rng(seed=42)
rng.binomial(n=10, p=0.5, size=3)       # Binomial(n, p): # of successes in n trials
rng.binomial(n=1, p=0.7, size=10)       # Bernoulli(p) = Binomial(n=1, p): 0/1 draws
rng.normal(loc=0, scale=1, size=5)      # Normal(mu, sigma): Gaussian draws
rng.poisson(lam=3, size=5)              # Poisson(lambda): # of events per interval
rng.uniform(low=0, high=1, size=5)      # Uniform[low, high): flat continuous draws
rng.choice([0, 1], p=[0.3, 0.7], size=10)  # sample values with given probabilities
rng.permutation(10)                     # random permutation of 0..9 (e.g., shuffling data)
```

### Inspect, index, reshape, combine, and summarize arrays

```python
arr = np.arange(12).reshape(3, 4)      # 3x4 array with values 0..11
arr.shape, arr.size, arr.dtype         # (3, 4) shape, 12 elements, int dtype
arr.astype(float)                      # convert dtype (e.g., int -> float)

arr[1, 2]                              # single element, row 1 column 2 (= 6)
arr[:, 1]                              # entire column 1 -> array([1, 5, 9])
arr[1:3, 1:3]                          # 2x2 block: rows 1-2, cols 1-2
mask = arr < 5                         # boolean mask, same shape as arr
arr[mask]                              # 1D array of values where mask is True

np.where(arr < 5, arr, -1)             # element-wise if/else: arr value if cond True, else -1
np.where(arr % 2 == 0, 'even', 'odd')  # cond can select any values (strings too)
np.where(mask)                         # indices where cond is True (tuple of index arrays, one per axis)
arr[np.where(mask)]                    # use those indices to index (same as arr[mask])

arr.reshape(2, 6)                      # same data, new shape (total size must match)
arr.T                                  # transpose (rows <-> columns)
arr.flatten()                          # copy as 1D array
np.concatenate((arr, arr), axis=0)     # stack vertically (axis=1 stacks horizontally)

arr + 1                                # element-wise arithmetic (broadcasting)
arr * 3                                # element-wise multiply (not matrix multiplication)
np.sqrt(arr)                           # element-wise functions: sqrt, exp, log, abs, ...
arr.mean(axis=0)                       # mean per column (axis=0 collapses rows)
arr.std(axis=1)                        # std per row (axis=1 collapses columns)
```

## 3. pandas: Load and Inspect

```python
import pandas as pd

# Import examples
# df = pd.read_csv('data.csv')
# df = pd.read_excel('data.xlsx')
# df = pd.read_sql(query, connection)
# df = pd.read_json('data.json')

df = pd.DataFrame({
    'country': ['DK', 'SE', 'DK'],
    'year': [2024, 2024, 2025],
    'revenue': [100, 120, 130],
    'profit': [10, -2, 16],
})

df.head()
df.shape
df.columns
df.dtypes
df.info()
df.describe()
df['country'].value_counts(dropna=False)
```

## 4. pandas: Select and Filter

```python
df['revenue']
df[['country', 'revenue']]

df.loc[df['country'] == 'DK', ['country', 'revenue']]
df.iloc[0, 1]

df[df['revenue'] > 100]
df[(df['revenue'] > 100) & (df['profit'] < 0)]
df[df['country'].isin(['DK', 'SE'])]
df[df['country'].notna()]

df.loc[df['country'] == 'DK', 'country'] = 'Denmark'
```

## 5. pandas: Clean and Transform

```python
df.isnull()
df.dropna()
df.fillna(0)
df['revenue'].astype(float)
df.rename(columns={'revenue': 'value'})
df.set_index('country')

df = df.rename(columns={'revenue': 'value'})
df['growth'] = df['value'].pct_change()
df['log_value'] = np.log(df['value'])
df = df.sort_values('value', ascending=False)

df['date'] = pd.to_datetime(['2024-01-01', '2024-02-01', '2025-01-01'])
df['year_from_date'] = df['date'].dt.year
df['month'] = df['date'].dt.month
```

## 6. pandas: Group, Aggregate, Reshape

```python
df.groupby('year')['value'].mean()

df.groupby('year').agg(
    mean_value=('value', 'mean'),
    total_profit=('profit', 'sum'),
    observations=('value', 'count'),
)

df.corr(numeric_only=True)

df.pivot_table(index='year', values=['value', 'profit'], aggfunc='mean')

# Combine datasets
# pd.concat([df1, df2], axis=0)
# df1.merge(df2, on='country', how='left')
# df1.join(df2, on='country', how='inner')
```

## 7. Visualize with pandas

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()

df.plot.line(x='year', y='value')
plt.show()

df.plot.scatter(x='year', y='value')
plt.show()

df['value'].plot.hist(bins=20)
plt.show()

df[['value', 'profit']].plot.box()
plt.show()
```

## 8. Matplotlib: Basic Charts

```python
x = [1, 2, 3, 4]
y = [4, 7, 5, 9]

plt.plot(x, y)
plt.scatter(x, y)
plt.bar(x, y)
plt.hist(y, bins=4)
plt.fill_between(x, [3, 5, 4, 7], [5, 9, 6, 11], alpha=0.2)
plt.show()

plt.plot(x, y, label='Series 1', color='blue', linestyle='--')
plt.legend(loc='upper left')
plt.grid(True)
plt.show()
```

## 9. Matplotlib: Labels and Layout

```python
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x, y, marker='o', label='Series')
ax.set_title('Title')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.legend()
ax.grid(True)
ax.axvline(x=2, color='red')
ax.axhline(y=6, color='green', linestyle='--')
ax.annotate('Point', xy=(2, 7), xytext=(2.5, 8))
fig.tight_layout()
plt.show()

fig, axes = plt.subplots(2, 2, figsize=(10, 7))
axes[0, 0].plot(x, y)
axes[0, 1].scatter(x, y)
axes[1, 0].hist(y)
axes[1, 1].bar(x, y)
fig.tight_layout()
```
